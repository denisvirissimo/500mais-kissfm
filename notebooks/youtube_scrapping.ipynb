```python
import pandas as pd
from ytmusicapi import YTMusic

def call(row):
    ytmusic = YTMusic()

    param = row['Artista'] + ' ' + row['Musica']
    print(param)
    result = ytmusic.search(param, filter='songs')

    if len(result) > 0:
        return result[0]['videoId']

df = pd.read_csv('./data/500+.csv')


newdf = df.groupby(['Artista', 'Musica', 'Observacao'], dropna=False)
newdf = newdf.size().reset_index(name='size').drop(columns='size')
newdf["SongID"] = (
    newdf["Artista"].astype(str).str.strip() + " — " +
    newdf["Musica"].astype(str).str.strip() + " — " +
    newdf["Observacao"].astype(str).str.strip()
)
newdf['YT'] = newdf.apply(call, axis=1)
newdf

newdf.to_csv('youtube.csv', index=False)
```
